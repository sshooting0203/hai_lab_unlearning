```python
# 1) domain -> intents 로딩
domain_map_path = "/dataset/hai_ssh/code/O3-LLM-UNLEARNING/ScienceQA_expriment/data/clinc150/CLINC_domain.json"
with open(domain_map_path, "r", encoding="utf-8") as f:
    domain2intents = json.load(f)
```

```python
# 2) intent -> domain (역매핑) 만들기
intent2domain = {}
dup = []
for domain, intents in domain2intents.items():
    for intent in intents:
        if intent in intent2domain and intent2domain[intent] != domain:
            dup.append((intent, intent2domain[intent], domain))
        intent2domain[intent] = domain
```

```python
df_train = df_train.rename(columns={0: "question", 1: "intent"})
df_val = df_val.rename(columns={0: "question", 1: "intent"})
df_test = df_test.rename(columns={0: "question", 1: "intent"})
```

```python
df_train["domain"] = df_train['intent'].map(intent2domain)
missing = df_train[df_train["domain"].isna()]['intent'].unique()
print("missing intents:", len(missing))
print(missing[:30])
df_train.head()
```

```python
df_test["domain"] = df_test['intent'].map(intent2domain)
missing = df_test[df_test["domain"].isna()]['intent'].unique()
print("missing intents:", len(missing))
print(missing[:30])
```

```python
out_dir = "/dataset/hai_ssh/code/O3-LLM-UNLEARNING/ScienceQA_expriment/data/clinc_text_12_intent"
os.makedirs(out_dir, exist_ok=True)

def save_domain_split_json(df, split_name: str, out_dir: str):
    if "domain" not in df.columns:
        raise ValueError(f"{split_name} dataframe has no 'domain' column")

    for domain, g in df.groupby("domain"):
        if domain is None or (isinstance(domain, float) and domain != domain):
            continue

        split_tag = "val" if split_name in ["val", "validation"] else split_name
        fp = os.path.join(out_dir, f"clincqa_{domain}_{split_tag}.json")
        columns_to_save = [c for c in ["question", "intent", "domain", "text_input", "text_full", "instruction", "input", "output"] if c in g.columns]
        records = g[columns_to_save].to_dict(orient="records")

        with open(fp, "w", encoding="utf-8") as f:
            json.dump(records, f, ensure_ascii=False, indent=2)

    print(f"[{split_name}] saved domain jsons into: {out_dir}")
```

```python
save_domain_split_json(df_train, "train", out_dir)
save_domain_split_json(df_val,   "validation",   out_dir)  
save_domain_split_json(df_test,  "test",  out_dir)
```

```python
import os, json
from datasets import load_dataset, DatasetDict

def save_like_commonqa(hf_name: str, out_root: str):
    ds_name = hf_name.split("/")[-1]
    out_dir = os.path.join(out_root, ds_name)
    os.makedirs(out_dir, exist_ok=True)

    ds = load_dataset(hf_name)  # 보통 DatasetDict
    if not isinstance(ds, DatasetDict):
        raise ValueError(f"{hf_name} did not return a DatasetDict. got type={type(ds)}")
    hf_path = os.path.join(out_dir, f"{ds_name}_all.hf")
    ds.save_to_disk(hf_path)
    for split in ds.keys():
        json_path = os.path.join(out_dir, f"{ds_name}_{split}.json")
        with open(json_path, "w", encoding="utf-8") as f:
            json.dump(ds[split].to_list(), f, ensure_ascii=False)
    print(f"[OK] saved: {hf_path}")
    print(ds)

save_like_commonqa("SetFit/mrpc", out_root="/dataset/hai_ssh/code/O3-LLM-UNLEARNING/ScienceQA_expriment/data/mrpc")
save_like_commonqa("SetFit/rte",  out_root="/dataset/hai_ssh/code/O3-LLM-UNLEARNING/ScienceQA_expriment/data/rte")
```
